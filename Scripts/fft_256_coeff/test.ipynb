```python
def indice_stage(n_stage, n_data):
    ans = []
    n_subdiv = n_data // (2**n_stage)
    for i in range(n_subdiv):
        for j in range(2**n_stage//2):
            ans.append(j*2+2**n_stage*i)
        for j in range(2**n_stage//2):
            ans.append(j*2+1+2**n_stage*i)
    return ans


print(indice_stage(2, 256))
print(indice_stage(4, 256))
print(indice_stage(7, 256))

```

```python
import numpy as np

def indice(stage, N):
    result = []
    m = 2 ** (stage + 1)
    half_m = m // 2        
    for k in range(0, N, m):
        for j in range(half_m):
            result.append((k+j, k+j+half_m))
    return result


print(indice(1, 256))
print(indice(2, 256))
print(indice(4, 256))
print(indice(7, 256))
```
